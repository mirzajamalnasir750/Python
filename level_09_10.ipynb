```python
students = [
    ("Ali", 45),
    ("Ahmed", 80),
    ("Sara", 65),
    ("Usman", 35),
    ("Ayesha", 90)
]

# Step 1: Sirf 50 ya us se zyada marks walay students
passed = filter(lambda x: x[1] >= 50, students)
```

```python

# Step 2: Har selected student ko 5 bonus marks
bonus = map(lambda x: (x[0], x[1] + 5), passed)

# Step 3: Highest marks pehle
result = sorted(bonus, key=lambda x: x[1], reverse=True)

print(result)
```

```python

# Level 9: Filter, Map and Sort

students = [
    ("Ali", 45),
    ("Ahmed", 80),
    ("Sara", 65),
    ("Usman", 35),
    ("Ayesha", 90)
]

passed = filter(lambda x: x[1] >= 50, students)
bonus = map(lambda x: (x[0], x[1] + 5), passed)
result = sorted(bonus, key=lambda x: x[1], reverse=True)

print("Level 9:", result)


# Level 10: Dictionary and Filter

students = {
    "Ali": 75,
    "Ahmed": 90,
    "Sara": 60,
    "Usman": 85
}

result = list(filter(lambda x: students[x] >= 80, students))

print("Level 10:", result)
```
