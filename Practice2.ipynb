```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

X_processed = new_dataset_Cat.copy()

# Age aur Weight ko 0–1 range mein scale karo
scaler = MinMaxScaler()

X_processed[["Weight (kg)", "Age (Years)"]] = scaler.fit_transform(
    X_processed[["Weight (kg)", "Age (Years)"]]
)

# Gender: Female = 0, Male = 1
X_processed["Gender"] = X_processed["Gender"].map({
    "Female": 0,
    "Male": 1
})

# Color ki One-Hot Encoding
X_processed = pd.get_dummies(
    X_processed,
    columns=["Color"],
    dtype=int
)

X_processed.head()
```

```python
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier,
    AdaBoostClassifier
)
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neural_network import MLPClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# Encoded target classes
classes = np.unique(np.concatenate([
    np.asarray(y_train).ravel(),
    np.asarray(y_test).ravel()
]))

# Original Breed names from your existing LabelEncoder
class_names = label_encoder.inverse_transform(
    classes.astype(int)
)

# Models: one by one run honge
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000, random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200, random_state=42, n_jobs=-1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=200, random_state=42, n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "AdaBoost": AdaBoostClassifier(
        random_state=42
    ),

    "SVM": SVC(kernel="rbf"),

    "KNN": KNeighborsClassifier(n_neighbors=5),

    "Gaussian Naive Bayes": GaussianNB(),

    "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),

    "SGD Classifier": SGDClassifier(
        loss="log_loss", max_iter=2000, random_state=42
    ),

    "Neural Network": MLPClassifier(
        hidden_layer_sizes=(64, 32),
        max_iter=1000,
        random_state=42
    )
}

trained_models = {}
predictions = {}
results = []
failures = []

for name, estimator in models.items():

    print(f"\nRunning: {name}")

    try:
        current_model = clone(estimator)
        start = time.perf_counter()

        # Train
        current_model.fit(X_train, y_train)

        # Predict
        train_pred = current_model.predict(X_train)
        test_pred = current_model.predict(X_test)

        train_accuracy = accuracy_score(y_train, train_pred)
        test_accuracy = accuracy_score(y_test, test_pred)

        precision, recall, f1, _ = precision_recall_fscore_support(
            y_test,
            test_pred,
            labels=classes,
            average="macro",
            zero_division=0
        )

        elapsed = time.perf_counter() - start

        trained_models[name] = current_model
        predictions[name] = test_pred

        results.append({
            "Model": name,
            "Train Accuracy (%)": train_accuracy * 100,
            "Test Accuracy (%)": test_accuracy * 100,
            "Macro Precision (%)": precision * 100,
            "Macro Recall (%)": recall * 100,
            "Macro F1 (%)": f1 * 100,
            "Time (seconds)": elapsed
        })

        print(f"Training accuracy: {train_accuracy:.2%}")
        print(f"Testing accuracy:  {test_accuracy:.2%}")
        print(f"Macro precision:   {precision:.2%}")
        print(f"Macro recall:      {recall:.2%}")
        print(f"Macro F1:          {f1:.2%}")

        # Confusion matrix
        cm = confusion_matrix(
            y_test, test_pred, labels=classes
        )

        fig, ax = plt.subplots(figsize=(13, 11))

        ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=class_names
        ).plot(
            ax=ax,
            cmap="Blues",
            xticks_rotation=90,
            include_values=True,
            values_format="d",
            colorbar=False
        )

        ax.tick_params(axis="both", labelsize=8)
        ax.set_title(f"{name} — Test Confusion Matrix")
        plt.tight_layout()
        plt.show()

        # Training vs testing accuracy
        fig, ax = plt.subplots(figsize=(6, 4))

        bars = ax.bar(
            ["Training", "Testing"],
            [train_accuracy * 100, test_accuracy * 100],
            color=["royalblue", "orange"]
        )

        ax.bar_label(bars, fmt="%.2f", padding=3)
        ax.set_ylim(0, 110)
        ax.set_ylabel("Accuracy (%)")
        ax.set_title(name)

        plt.tight_layout()
        plt.show()

    except Exception as error:
        failures.append({
            "Model": name,
            "Error": str(error)
        })
        print(f"{name} failed: {error}")


# Final comparison
results_df = pd.DataFrame(results)

if not results_df.empty:

    results_df = results_df.sort_values(
        "Macro F1 (%)", ascending=False
    ).reset_index(drop=True)

    display(results_df.round(2))

    ax = results_df.set_index("Model")[
        ["Train Accuracy (%)", "Test Accuracy (%)"]
    ].plot.bar(
        figsize=(15, 6),
        color=["royalblue", "orange"]
    )

    for bars in ax.containers:
        ax.bar_label(bars, fmt="%.1f", padding=3, fontsize=8)

    ax.set_ylim(0, 110)
    ax.set_ylabel("Accuracy (%)")
    ax.set_title("All Models — Training vs Testing Accuracy")

    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

if failures:
    print("\nModels that failed:")
    display(pd.DataFrame(failures))
```
