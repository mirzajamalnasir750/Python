## 1. Quick refresh: Python lists vs NumPy arrays

A Python list can hold mixed types. A NumPy array is designed for efficient numerical operations and usually stores one data type.

```python
python_list = [1, 2, 3, 4]
numpy_array = np.array([1, 2, 3, 4])

print("List doubled:", python_list * 2)       # repeats the list
print("Array doubled:", numpy_array * 2)      # multiplies each number
```

### Create arrays

```python
a = np.array([10, 20, 30, 40])
zeros = np.zeros(4)
ones = np.ones((2, 3))
sequence = np.arange(0, 10, 2)
evenly_spaced = np.linspace(0, 1, 5)

print(a)
print(zeros)
print(ones)
print(sequence)
print(evenly_spaced)
```

## 2. NumPy shapes, dimensions, and data types

```python
one_d = np.array([1, 2, 3])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
three_d = np.ones((2, 3, 4))

print("1D shape:", one_d.shape, "dimensions:", one_d.ndim)
print("2D shape:", two_d.shape, "dimensions:", two_d.ndim)
print("3D shape:", three_d.shape, "dimensions:", three_d.ndim)
print("Data type:", two_d.dtype)
print("Total elements:", two_d.size)
```

### Convert data types

```python
values = np.array([1.2, 2.8, 3.5])
print(values.astype(int))  # truncates decimal parts; it does not round
```

## 3. Indexing and slicing

```python
arr = np.array([5, 10, 15, 20, 25, 30])
print(arr[0])       # first item
print(arr[-1])      # last item
print(arr[1:4])     # index 1, 2, 3
print(arr[::2])     # every second item
print(arr[::-1])    # reverse
```

### 2D indexing and slicing

```python
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print(matrix[0, 1])    # row 0, column 1 -> 20
print(matrix[1, :])    # second row
print(matrix[:, 0])    # first column
print(matrix[:2, 1:])  # first two rows, columns 1 onward
```

## 4. Reshaping, flattening, and transposing

```python
numbers = np.arange(1, 13)
grid = numbers.reshape(3, 4)

print(grid)
print("Transpose:\n", grid.T)
print("Flatten copy:", grid.flatten())
print("Ravel view when possible:", grid.ravel())
```

## 5. Vectorised arithmetic and universal functions

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])

print(x + y)
print(y - x)
print(x * y)
print(y / x)
print(x ** 2)
print(np.sqrt(x))
print(np.exp(x))
print(np.log(x))
```

## 6. Boolean masks and filtering

```python
scores = np.array([45, 72, 88, 59, 95, 67])
passed = scores >= 60

print("Mask:", passed)
print("Passed scores:", scores[passed])
print("Scores from 60 to 80:", scores[(scores >= 60) & (scores <= 80)])
print("Failed scores:", scores[scores < 60])
```

**Important:** Use `&` for element-wise AND and `|` for element-wise OR with NumPy arrays. Put each condition in parentheses. Do not use Python's `and` / `or` for array masks.

## 7. Aggregations and axis

```python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("Total:", data.sum())
print("Mean:", data.mean())
print("Column sums (axis=0):", data.sum(axis=0))
print("Row sums (axis=1):", data.sum(axis=1))
print("Column means:", data.mean(axis=0))
print("Min / max:", data.min(), data.max())
print("Standard deviation:", data.std())
```

## 8. Broadcasting

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])
row_values = np.array([10, 20, 30])

print(matrix + row_values)  # row_values is broadcast across rows
print(matrix * 10)          # scalar broadcasts to every element
```

## 9. Sorting, unique values, and counts

```python
values = np.array([4, 2, 4, 1, 2, 4, 3])
print("Sorted:", np.sort(values))
print("Unique:", np.unique(values))
print("Unique + counts:", np.unique(values, return_counts=True))
```

## 10. Random numbers and reproducibility

```python
rng = np.random.default_rng(42)  # fixed seed for reproducible examples
print(rng.integers(1, 11, size=5))      # integers 1 through 10
print(rng.normal(loc=0, scale=1, size=5))
print(rng.choice(["A", "B", "C"], size=5))
```

## 11. Linear algebra basics

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Element-wise multiplication:\n", A * B)
print("Matrix multiplication:\n", A @ B)
print("Transpose:\n", A.T)
print("Determinant:", np.linalg.det(A))
print("Inverse:\n", np.linalg.inv(A))
```

Use `np.linalg.solve(A, b)` to solve a linear system `A x = b`. In practice, avoid computing an inverse just to solve a system.

```python
A = np.array([[3, 1], [1, 2]], dtype=float)
b = np.array([9, 8], dtype=float)
solution = np.linalg.solve(A, b)
print("Solution:", solution)
print("Check A @ solution:", A @ solution)
```

## 12. NumPy missing and special values

```python
vals = np.array([1.0, 2.0, np.nan, 4.0])
print("NaN mask:", np.isnan(vals))
print("Mean ignoring NaN:", np.nanmean(vals))
print("Is finite:", np.isfinite(vals))
```
