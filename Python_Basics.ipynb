```python

def second_largest(numbers):
    largest = float('-inf')
    second = float('-inf')

    for num in numbers:
        if num > largest:
            second = largest
            largest = num
        elif num < largest and num > second:
            second = num

    if second == float('-inf'):
        return None

    return second


numbers = [10, 5, 20, 8, 20, 15]

second_largest_number = second_largest(numbers)

print("Second largest number:", second_largest_number)
```

```python
def separate_unique_items(items):
    numbers = []
    characters = []
    symbols = []

    for item in items:
        if isinstance(item, (int, float)):
            if item not in numbers:
                numbers.append(item)

        elif isinstance(item, str) and item.isalpha():
            if item not in characters:
                characters.append(item)

        else:
            if item not in symbols:
                symbols.append(item)

    return numbers, characters, symbols


items = [1, 2, 3, 2, 4, 'a', 'b', 'a', '@', '#', '$', '@', '%', '!']

numbers, characters, symbols = separate_unique_items(items)

print("Numbers:", numbers)
print("Characters:", characters)
print("Symbols:", symbols)
```

```python
def separate_unique_items(items):
    numbers = []
    strings = []
    characters = []
    symbols = []

    for item in items:
        if isinstance(item, (int, float)):
            if item not in numbers:
                numbers.append(item)

        elif isinstance(item, str):
            if item.isalpha():
                if len(item) == 1:
                    if item not in characters:
                        characters.append(item)
                else:
                    if item not in strings:
                        strings.append(item)

            else:
                if item not in symbols:
                    symbols.append(item)

    return numbers, strings, characters, symbols


items = [1, 2, 3, 'apple', 'hello', 'a', 'b', 'apple',
         '@', '#', 'hello', 'cat', 'x']

numbers, strings, characters, symbols = separate_unique_items(items)

print("Numbers:", numbers)
print("Strings:", strings)
print("Characters:", characters)
print("Symbols:", symbols)
```

```python

def check_palindrome(items):
    for item in items:
        if item == item[::-1]:
            print(item, "is palindrome")
        else:
            print(item, "is not palindrome")


items = ['a', 'aa', 'aaa', 'abab', 'abba', 'accb']

check_palindrome(items)
```

```python
items = ['a', 'aa', 'aaa', 'abab', 'abba', 'accb']

def find_by_start(items, character):
    result = []

    for item in items:
        if item.startswith(character):
            result.append(item)

    return result

print(find_by_start(items, 'a'))
```

```python
items = ['a', 'aa', 'aaa', 'abab', 'abba', 'accb']

def find_by_end(items, character):
    result = []

    for item in items:
        if item.endswith(character):
            result.append(item)

    return result

print(find_by_end(items, 'a'))
```
